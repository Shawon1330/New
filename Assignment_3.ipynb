**Python Programming Assignments-Conditional**

Assignment 1: Basic If-Else Statement

```python
age = int(input("Enter your age: "))

if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are NOT eligible to vote.")
```

Assignment 2: Nested If-Else

```python
num = float(input("Enter a number: "))

if num > 0:
    print("Positive Number")
elif num < 0:
    print("Negative Number")
else:
    print("Zero")
```

Assignment 3: Conditional Expressions

```python
number = int(input("Enter a number: "))

print("Even Number" if number % 2 == 0 else "Odd Number")
```

Assignment 4: Age Group Classification

```python
age = int(input("Enter your age: "))

if 0 <= age <= 12:
    print("Child")
elif 13 <= age <= 19:
    print("Teenager")
elif 20 <= age <= 64:
    print("Adult")
elif age >= 65:
    print("Senior")
else:
    print("Invalid age")
```

Python Programming Assignments - Loops and Functions

Assignment 1: Sum of Natural Numbers

```python
def sum_of_natural_numbers(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

print("Sum:", sum_of_natural_numbers(10))
```

Assignment 2: Factorial of a Number

```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print("Factorial:", factorial(5))
```

Assignment 3: Multiplication Table

```python
def multiplication_table(n):
    for i in range(1, 11):
        print(f"{n} x {i} = {n * i}")

multiplication_table(5)
```
