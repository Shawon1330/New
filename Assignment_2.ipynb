```python
# 8. Create two sets of colors and find their union and intersection.


set1 = {"red", "green", "blue", "yellow"}
set2 = {"blue", "black", "white", "green"}


union_set = set1.union(set2)
intersection_set = set1.intersection(set2)

print("Union:", union_set)
print("Intersection:", intersection_set)
```

```python
# 9. Write a Python function that takes a string as input and prints the first and last character of the string.

def first_last_char(text):
    if len(text) == 0:
        print("Empty string!")
    else:
        print("First character:", text[0])
        print("Last character:", text[-1])

first_last_char("Shawon")
```

```python
# 10. Write a Python function that takes a string and converts it to uppercase.


def to_uppercase(text):
    result = text.upper()
    print(result)
to_uppercase("machine learning")
```
